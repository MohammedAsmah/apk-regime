

*  **On prend le nom du plat détecté par l’IA (nom_du_plat)
* **On cherche un fichier JSON correspondant dans ton dossier**.



* ** Si le fichier existe, on lit les valeurs nutritionnelles et on les affiches**
*   **Si le plat n’existe pas, on affiche un message d’erreur **

.

```python
import os
import json
import unicodedata
import re

# CONFIGURATION

DOSSIER_JSON = "/content/drive/MyDrive/data/data"

# Résultats du modèle IA
classe = "rice"
nom_du_plat ="Rice Krispies"   # exemple nom  détecté par l'IA
```

```python
# ===============================
# FONCTION DE NORMALISATION
# ===============================
def normaliser(texte):
    texte = texte.lower()
    texte = unicodedata.normalize("NFD", texte)
    texte = "".join(c for c in texte if unicodedata.category(c) != "Mn")
    texte = re.sub(r"[^a-z0-9 ]+", "", texte)
    return texte.strip()
```

```python


# ===============================
# CHARGER LE FICHIER DE CLASSE
# ===============================
fichier_classe = os.path.join(DOSSIER_JSON, f"{classe}.json")

if not os.path.exists(fichier_classe):
    print(f"❌ Classe inconnue : {classe}")
    exit()


with open(fichier_classe, "r", encoding="utf-8") as f:
    data = json.load(f)

if not isinstance(data, list):
    print("❌ Format JSON invalide (liste attendue)")
    exit()

```

```python
import unicodedata
import re

def normaliser(texte):
    texte = texte.lower()
    texte = unicodedata.normalize("NFD", texte)
    texte = "".join(c for c in texte if unicodedata.category(c) != "Mn")
    texte = re.sub(r"[^a-z0-9 ]+", "", texte)
    return texte.strip()

```

```python
nom_recherche = normaliser(nom_du_plat)

plat_trouve = None
for plat in data:
    nom_plat_json = normaliser(plat.get("name", ""))
    if nom_recherche in nom_plat_json:
        plat_trouve = plat
        break
```

```python
# ===============================
# AFFICHAGE
# ===============================
if plat_trouve:
    print("✅ Classe :", classe)
    print("🍽 Plat :", plat_trouve.get("name", "N/A"))
    print("Calories :", plat_trouve.get("calories_kcal", "N/A"), "kcal")
    print("Protéines :", plat_trouve.get("protein_g", "N/A"), "g")
    print("Lipides :", plat_trouve.get("fat_g", "N/A"), "g")
    print("Glucides :", plat_trouve.get("carbs_g", "N/A"), "g")
    print("Sucres :", plat_trouve.get("sugars_g", "N/A"), "g")
    print("Fibres :", plat_trouve.get("fiber_g", "N/A"), "g")
    print("Sodium :", plat_trouve.get("sodium_mg", "N/A"), "mg")
    print("Portion :", plat_trouve.get("serving_size_g", "N/A"), "g")
else:
    print(f"❌ Plat '{nom_du_plat}' non trouvé dans la classe '{classe}'")
```

**Calories pour une portion personnalisée**

```python
def calcul_par_portion(valeur, portion_reference, portion_user):
    return (valeur / portion_reference) * portion_user

portion_user = 45  # g mangés par l'utilisateur

calories = calcul_par_portion(389, 30, portion_user)
print("Calories estimées :", round(calories, 1), "kcal")
```

Répartition macro (%)

```python
total_macros = 7 + 1.2 + 86

print("Protéines :", round(7 / total_macros * 100, 1), "%")
print("Lipides :", round(1.2 / total_macros * 100, 1), "%")
print("Glucides :", round(86 / total_macros * 100, 1), "%")
```

**Densité calorique (kcal / 100g)**

```python
kcal_100g = (389 / 30) * 100
print("Densité calorique :", round(kcal_100g), "kcal / 100g")
```
